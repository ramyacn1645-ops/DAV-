```python
sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df,x="SepalLengthCm",bins=10,color="yellow")
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Number of Flowers")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Species",y="SepalLengthCm",color="pink")
plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Lengt (cm)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="SepalLengthCm",y="PetalLengthCm",hue="Species")
plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
sns.pairplot(df,hue="Species")
plt.show()
```

```python
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]
correlation = df[numeric_columns].corr()
print(correlation)
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
